```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
import plotly.express as px
import textwrap  # <--- Importante para dividir el texto en renglones
from plotly.colors import sample_colorscale

sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.sans-serif': 'DejaVu Sans', 'font.size': 11})

df_analitico = pd.read_parquet("../data/processed/sequia_analitico_2016.parquet")

df_analitico['fecha'] = pd.to_datetime(df_analitico['fecha'])
df_analitico['anio'] = df_analitico['fecha'].dt.year

cat_order = ['Sin Sequía', 'D0', 'D1', 'D2', 'D3', 'D4']
```

```python
df_analitico['fecha'] = pd.to_datetime(df_analitico['fecha'])
df_analitico['anio'] = df_analitico['fecha'].dt.year

mapa_intensidad = {1: 'D0 Anormalmente seco', 2: 'D1 Sequía moderada', 3: 'D2 Sequía severa', 4: 'D3 Sequía extrema', 5: 'D4 Sequía excepcional'}
orden_niveles = list(mapa_intensidad.values())

df_filtrado = df_analitico[df_analitico['intensidad'].isin(mapa_intensidad)].copy()
df_filtrado['intensidad_nom'] = df_filtrado['intensidad'].map(mapa_intensidad)

df_g2 = (
    df_filtrado.groupby(['anio', 'intensidad_nom'])['CVE_CONCATENADA']
    .nunique()
    .reset_index(name='municipios')
)

# Porcentaje respecto al total del año
total_por_anio = df_analitico.groupby('anio')['CVE_CONCATENADA'].nunique().rename('total_anio')
df_g2 = df_g2.merge(total_por_anio, on='anio')
df_g2['pct'] = (df_g2['municipios'] / df_g2['total_anio'] * 100).round(1)

# Año incompleto 2026
ultimo_anio = df_g2['anio'].max()
df_g2['anio_txt'] = df_g2['anio'].astype(str)
df_g2.loc[df_g2['anio'] == ultimo_anio, 'anio_txt'] = f'{ultimo_anio}*'
orden_anios = sorted(df_g2['anio_txt'].unique())

# Color por intensidad
colores_sequia = {
    'D0 Anormalmente seco':  '#FFFF00',
    'D1 Sequía moderada':    '#FCD37F',
    'D2 Sequía severa':      '#FFAA00',
    'D3 Sequía extrema':     '#E60000',
    'D4 Sequía excepcional': '#730000',
}

fig = px.bar(
    df_g2,
    x='anio_txt',
    y='municipios',
    color='intensidad_nom',
    barmode='group',
    category_orders={'intensidad_nom': orden_niveles, 'anio_txt': orden_anios},
    color_discrete_map=colores_sequia,
    custom_data=['intensidad_nom', 'pct', 'total_anio'],
    labels={
        'anio_txt': 'Año',
        'municipios': 'Número de municipios afectados',
        'intensidad_nom': 'Intensidad de sequía'
    }
)

fig.update_traces(
    marker_line_color='rgba(0,0,0,0.35)',
    marker_line_width=0.6,
    hovertemplate='<b>Año %{x}</b><br>'
                  'Nivel: %{customdata[0]}<br>'
                  'Municipios afectados: %{y:,}<br>'
                  '%{customdata[1]}% de %{customdata[2]:,} evaluados'
                  '<extra></extra>'
)

fig.update_layout(
    title=dict(
        text='<b>Municipios afectados por intensidad de sequía, por año</b>',
        font=dict(size=16)
    ),
    template='plotly_white',
    bargap=0.15,
    bargroupgap=0.04,
    legend=dict(title='Nivel', orientation='v', yanchor='top', y=1, xanchor='left', x=1.02),
    height=520,
    margin=dict(l=60, r=30, t=80, b=90),
)
```

```python
# GRAFICA 3: Barras apiladas interactivas (con Plotly)
df_analitico['fecha'] = pd.to_datetime(df_analitico['fecha'])

mapa_intensidad = {
    0: 'Sin sequía',
    1: 'D0 Anormalmente seco',
    2: 'D1 Sequía moderada',
    3: 'D2 Sequía severa',
    4: 'D3 Sequía extrema',
    5: 'D4 Sequía excepcional'
}
orden_niveles = list(mapa_intensidad.values())

df_analitico['intensidad_num'] = pd.to_numeric(df_analitico['intensidad'], errors='coerce').fillna(0).astype(int)
df_max_estado = df_analitico.groupby(['fecha', 'ENTIDAD'])['intensidad_num'].max().reset_index()

df_max_estado['categoria_maxima'] = df_max_estado['intensidad_num'].map(mapa_intensidad)
df_max_estado['categoria_maxima'] = pd.Categorical(df_max_estado['categoria_maxima'], categories=orden_niveles, ordered=True)

# Obtenemos el número de estados
df_conteo = df_max_estado.groupby(['fecha', 'categoria_maxima'])['ENTIDAD'].nunique().reset_index(name='num_estados')

def formatear_lista_estados(estados):
    texto_unido = ', '.join(sorted(estados))
    renglones = textwrap.wrap(texto_unido, width=50) # Ajusta 'width' si lo quieres más ancho o angosto
    return '<br>'.join(renglones)

df_nombres = df_max_estado.groupby(['fecha', 'categoria_maxima'])['ENTIDAD'].apply(
    formatear_lista_estados
).reset_index(name='lista_estados')

# Unimos ambos datos
df_plotly = pd.merge(df_conteo, df_nombres, on=['fecha', 'categoria_maxima'])
df_plotly = df_plotly[df_plotly['num_estados'] > 0]

colores_custom_plotly = {
    'Sin sequía':            '#2ca02c',
    'D0 Anormalmente seco':  '#FFFF00',
    'D1 Sequía moderada':    '#FCD37F',
    'D2 Sequía severa':      '#FFAA00',
    'D3 Sequía extrema':     '#E60000',
    'D4 Sequía excepcional': '#730000',
}

fig = px.bar(
    df_plotly,
    x='fecha',
    y='num_estados',
    color='categoria_maxima',
    custom_data=['lista_estados'],
    category_orders={"categoria_maxima": cat_order},
    color_discrete_map=colores_custom_plotly
)

# %{x|%d-%m-%Y} = Toma el eje X (fecha) y le da formato Día-Mes-Año
fig.update_traces(
    hovertemplate=(
        "<b>Fecha: %{x|%d-%m-%Y}</b><br>"
        "<b>Nivel: %{data.name}</b><br>"
        "Total de Estados: %{y}<br>"
        "Entidades:<br>%{customdata[0]}"
        "<extra></extra>"
    )
)

fig.update_layout(
    title=dict(
        text='<b>Distribución temporal de estados por nivel máximo de sequía</b>',
        font=dict(size=16)
    ),
    template='plotly_white',
    barmode='stack',
    bargap=0,
    hovermode='closest',
    legend=dict(title='Intensidad de sequía', orientation='v', yanchor='top', y=1, xanchor='left', x=1.02),
    height=540
)

fig.update_xaxes(tickformat="%d-%m-%Y", showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor='lightgray')

fig.show()
```

```python
# GRAFICA 4.1: Estados comparados en una fecha de referencia (% de municipios por nivel)
df_analitico['fecha'] = pd.to_datetime(df_analitico['fecha'])

# Fecha de referencia: la más reciente. Cámbiala si quieres otra, ej. pd.Timestamp('2024-05-15')
fecha_ref = df_analitico['fecha'].max()

mapa_intensidad = {
    0: 'Sin sequía',
    1: 'D0 Anormalmente seco',
    2: 'D1 Sequía moderada',
    3: 'D2 Sequía severa',
    4: 'D3 Sequía extrema',
    5: 'D4 Sequía excepcional'
}
orden_niveles = list(mapa_intensidad.values())

df_ref = df_analitico[df_analitico['fecha'] == fecha_ref].copy()
df_ref['intensidad'] = df_ref['intensidad'].fillna(0)
df_ref['nivel'] = df_ref['intensidad'].map(mapa_intensidad)

# Municipios por estado y nivel
df_g4 = (
    df_ref.groupby(['ENTIDAD', 'nivel'])['CVE_CONCATENADA']
    .nunique()
    .reset_index(name='municipios')
)
total_mun = df_g4.groupby('ENTIDAD')['municipios'].sum().rename('total_mun')
df_g4 = df_g4.merge(total_mun, on='ENTIDAD')
df_g4['pct'] = (df_g4['municipios'] / df_g4['total_mun'] * 100).round(1)

# Orden de los estados: % de municipios en D2 o peor (el más crítico queda arriba)
pct_d2_mas = df_ref.groupby('ENTIDAD')['intensidad'].apply(lambda x: (x >= 3).mean() * 100)
orden_estados = pct_d2_mas.sort_values(ascending=True).index.tolist()

colores = {
    'Sin sequía':            '#2ca02c',
    'D0 Anormalmente seco':  '#FFFF00',
    'D1 Sequía moderada':    '#FCD37F',
    'D2 Sequía severa':      '#FFAA00',
    'D3 Sequía extrema':     '#E60000',
    'D4 Sequía excepcional': '#730000',
}

fig = px.bar(
    df_g4,
    x='pct',
    y='ENTIDAD',
    color='nivel',
    orientation='h',
    barmode='stack',
    category_orders={'nivel': orden_niveles, 'ENTIDAD': orden_estados},
    color_discrete_map=colores,
    custom_data=['nivel', 'municipios', 'total_mun'],
    labels={'pct': '% de municipios del estado', 'ENTIDAD': 'Estado', 'nivel': 'Nivel de sequía'}
)

fig.update_traces(
    marker_line_color='rgba(0,0,0,0.25)',
    marker_line_width=0.4,
    hovertemplate='<b>%{y}</b><br>'
                  '%{customdata[0]}<br>'
                  '%{customdata[1]:,} de %{customdata[2]:,} municipios (%{x}%)'
                  '<extra></extra>'
)

fig.update_layout(
    title=dict(
        text=f'<b>Municipios por nivel de sequía en cada estado</b><br>'
             f'<span style="font-size:13px;color:gray">Corte al {fecha_ref:%d/%m/%Y} · ordenado por % de municipios en D2 o peor</span>',
        font=dict(size=16)
    ),
    template='plotly_white',
    height=820,
    legend=dict(title='Nivel de sequía', orientation='v', yanchor='top', y=1, xanchor='left', x=1.02),
    margin=dict(l=140, r=30, t=100, b=60)
)
fig.update_xaxes(range=[0, 100], ticksuffix='%', showgrid=True, gridcolor='lightgray')
fig.update_yaxes(showgrid=False)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# 1. Cargar datos
df_mun = pd.read_parquet("../data/processed/resumen_municipal.parquet")
df_mun['Municipio_Estado'] = df_mun['NOMBRE_MUN'] + ', ' + df_mun['ENTIDAD']

# 2. Extraer el Top 15 basado en la Frecuencia (en lugar de la racha)
top = (
    df_mun.nlargest(15, 'frecuencia_sequia_pct', keep='all')
    .sort_values('frecuencia_sequia_pct', ascending=True) # Ascendente para que el mayor quede hasta arriba en la gráfica
    .copy()
)

# 3. Construir la gráfica cruzando Frecuencia (eje X) e Intensidad (Color)
fig = px.bar(
    top,
    x='frecuencia_sequia_pct',
    y='Municipio_Estado',
    orientation='h',
    color='intensidad_media', 
    color_continuous_scale='Reds',
    text='frecuencia_sequia_pct',
    custom_data=['intensidad_media', 'racha_maxima_quincenas'],
    labels={
        'frecuencia_sequia_pct': '% de tiempo en sequía (Frecuencia)',
        'Municipio_Estado': 'Municipio',
        'intensidad_media': 'Intensidad Media'
    }
)

# 4. Ajustar el formato del texto y la tarjeta de información (Hover)
fig.update_traces(
    texttemplate='%{text:.1f}%', # Muestra el porcentaje con 1 decimal
    textposition='outside',
    hovertemplate='<b>%{y}</b><br>'
                  'Frecuencia histórica: %{x:.1f}% del tiempo<br>'
                  'Intensidad promedio: %{customdata[0]:.2f} (Escala D0-D4)<br>'
                  'Racha máxima: %{customdata[1]} quincenas seguidas'
                  '<extra></extra>'
)

# 5. Formato ejecutivo
fig.update_layout(
    title=dict(
        text='<b>Municipios con Mayor Frecuencia e Intensidad de Sequía Histórica</b><br>',
        font=dict(size=16)
    ),
    template='plotly_white',
    height=620,
    margin=dict(l=20, r=50, t=100, b=60),
    coloraxis_colorbar=dict(title='Intensidad<br>Media')
)

# Ampliar ligeramente el eje X al 110% para que los textos externos no se corten
fig.update_xaxes(showgrid=True, gridcolor='lightgray', rangemode='tozero', range=[0, 110])

fig.show()
fig.write_html('grafica_municipios_cronicos.html', include_plotlyjs='cdn')
```

```python
#| code-fold: true
#| code-summary: "Ver código de preprocesamiento y entrenamiento K-Means"

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans
import plotly.graph_objects as go

df = pd.read_parquet("../data/processed/resumen_municipal.parquet")

features = [
    'frecuencia_sequia_pct', 
    'racha_maxima_quincenas',
    'vulnerabilidad_ambiental', 
    'vulnerabilidad_social', 
    'vulnerabilidad_economica'
]

imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(df[features])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster_ID'] = kmeans.fit_predict(X_scaled)

cluster_risk = pd.DataFrame(X_scaled).groupby(df['Cluster_ID']).mean().sum(axis=1).sort_values()
mapeo_nombres = {
    cluster_risk.index[0]: 'Resilientes',
    cluster_risk.index[1]: 'Prioritarios',
    cluster_risk.index[2]: 'Intervención Crítica',
}

df['Perfil_Riesgo'] = df['Cluster_ID'].map(mapeo_nombres)

df.to_parquet("../data/processed/municipios_clusterizados.parquet", index=False)

print("Distribución de municipios por perfil de riesgo:")
df['Perfil_Riesgo'].value_counts()

perfiles = df.groupby('Perfil_Riesgo')[features].mean().reset_index()

for col in features:
    min_val = df[col].min()
    max_val = df[col].max()
    perfiles[col + '_norm'] = (perfiles[col] - min_val) / (max_val - min_val)

features_norm = [f + '_norm' for f in features]
nombres_ejes = ['Frecuencia Sequía', 'Racha Máxima', 'Vul. Ambiental', 'Vul. Social', 'Vul. Económica']

colores_cluster = {
    'Resilientes': '#c0a900',
    'Prioritarios': '#eb5e5e', 
    'Intervención Crítica': '#691edb',         
}

fig = go.Figure()

for i, row in perfiles.iterrows():
    perfil = row['Perfil_Riesgo']
    valores = row[features_norm].tolist()
    valores += [valores[0]]
    ejes_cerrados = nombres_ejes + [nombres_ejes[0]]
    
    fig.add_trace(go.Scatterpolar(
        r=valores,
        theta=ejes_cerrados,
        fill='toself',
        name=perfil,
        line_color=colores_cluster[perfil],
        opacity=0.7
    ))

fig.update_layout(
    polar=dict(
        radialaxis=dict(visible=False, range=[0, 1])
    ),
    showlegend=True,
    title='<b>Perfiles Territoriales: Descomposición del Riesgo</b>',
    template='plotly_white',
    legend=dict(orientation="h", y=-0.2)
)

fig.show()
```

```python
#| code-fold: true
#| code-summary: "Ver código de validación del modelo (Método del Codo)"

import pandas as pd
import plotly.graph_objects as go
from sklearn.cluster import KMeans
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

df = pd.read_parquet("../data/processed/resumen_municipal.parquet")
features = ['frecuencia_sequia_pct', 'racha_maxima_quincenas', 
            'vulnerabilidad_ambiental', 'vulnerabilidad_social', 'vulnerabilidad_economica']

X = SimpleImputer(strategy='median').fit_transform(df[features])
X_scaled = StandardScaler().fit_transform(X)

inercia = []
rango_k = range(2, 11)

for k in rango_k:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inercia.append(kmeans.inertia_)

fig_codo = go.Figure()
fig_codo.add_trace(go.Scatter(
    x=list(rango_k), 
    y=inercia, 
    mode='lines+markers',
    marker=dict(size=10, color='#b05c38'), 
    line=dict(width=3)
))

fig_codo.add_vline(x=3, line_width=2, line_dash="dash", line_color="gray")


fig_codo.update_layout(
    title='<b>Validación: Método del Codo (Inercia)</b>',
    xaxis_title='Número de Segmentos (k)',
    yaxis_title='Inercia Intra-Clúster ($WCSS$)',
    template='plotly_white',
    margin=dict(l=20, r=20, t=50, b=20)
)

fig_codo.show()
```
