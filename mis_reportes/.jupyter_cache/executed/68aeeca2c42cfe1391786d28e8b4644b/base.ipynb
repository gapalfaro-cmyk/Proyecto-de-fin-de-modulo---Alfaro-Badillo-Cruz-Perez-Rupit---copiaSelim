```python
#| label: setup

import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

pd.set_option("display.max_columns", 10)

RUTA_DATOS = "../data/raw/MunicipiosSequiaCSV.csv"

ID_COLS = [
    "CVE_CONCATENADA", "CVE_ENT", "CVE_MUN", "NOMBRE_MUN", "ENTIDAD",
    "ORG_CUENCA*", "CLV_OC", "CON_CUENCA", "CVE_CONC",
]

MESES = {
    "ene": 1, "feb": 2, "mar": 3, "abr": 4, "may": 5, "jun": 6,
    "jul": 7, "ago": 8, "sep": 9, "oct": 10, "nov": 11, "dic": 12,
}

# Nivel ordinal de severidad: sin sequía < D0 < D1 < D2 < D3 < D4
NIVEL_ORDINAL = {np.nan: 0, "D0": 1, "D1": 2, "D2": 3, "D3": 4, "D4": 5}
```

```python
#| label: carga

df = pd.read_csv(RUTA_DATOS, encoding="utf-8-sig", low_memory=False)
df.columns = [c.strip() for c in df.columns]

date_cols = [c for c in df.columns if c not in ID_COLS]
n_municipios = df.shape[0]
n_entidades = df["ENTIDAD"].nunique()
n_fechas = len(date_cols)

print(f"Municipios: {n_municipios:,}")
print(f"Columnas de identificación: {len(ID_COLS)}")
print(f"Columnas de fecha (periodos publicados): {n_fechas}")
print(f"Entidades federativas: {n_entidades}")
```

```python
#| label: integridad

duplicados = df["CVE_CONCATENADA"].duplicated().sum()
nulos_id = df[ID_COLS].isna().sum().sum()

print(f"Claves de municipio duplicadas: {duplicados}")
print(f"Valores nulos en columnas de identificación: {nulos_id}")

assert duplicados == 0, "Hay CVE_CONCATENADA duplicadas"
assert nulos_id == 0, "Hay nulos en columnas de identificación"
```

```python
#| label: fig-vacios-por-fecha
#| fig-cap: Porcentaje de municipios sin categoría asignada, por fecha de publicación

frac_vacio = df[date_cols].isna().mean()
cols_missing_real = frac_vacio[frac_vacio == 1.0].index.tolist()

fig, ax = plt.subplots()
ax.plot(range(len(frac_vacio)), frac_vacio.values * 100, linewidth=0.8)
ax.set_xlabel("Corte de publicación (orden cronológico)")
ax.set_ylabel("% de municipios vacíos")
ax.axhline(100, color="firebrick", linestyle="--", linewidth=0.8, label="100% vacío = hueco real")
ax.legend()
plt.tight_layout()
plt.show()

print(f"Fechas con 100% de celdas vacías (hueco real de publicación): {cols_missing_real}")
```

```python
#| label: fn-parseo

def parsear_fecha(col_name: str) -> pd.Timestamp:
    """Convierte 'dd-mmm-aa' (español) a Timestamp, p.ej. '15-sep-26' -> 2026-09-15."""
    dia, mes_abr, anio2 = col_name.split("-")
    return pd.Timestamp(year=2000 + int(anio2), month=MESES[mes_abr.lower()], day=int(dia))
```

```python
#| label: wide-to-long

largo = df.melt(id_vars=ID_COLS, value_vars=date_cols,
                 var_name="fecha_str", value_name="categoria")

largo["fecha"] = largo["fecha_str"].map(parsear_fecha)
largo["es_hueco_publicacion"] = largo["fecha_str"].isin(cols_missing_real)

# Nivel ordinal de severidad (0-5); NaN real para huecos de publicación
largo["nivel_severidad"] = largo["categoria"].map(NIVEL_ORDINAL)
largo.loc[largo["es_hueco_publicacion"], "nivel_severidad"] = np.nan

largo["categoria"] = largo["categoria"].fillna("Sin sequía")
largo.loc[largo["es_hueco_publicacion"], "categoria"] = np.nan

# Periodicidad de la fuente: mensual hasta 2013, quincenal desde 2014
largo["periodicidad"] = np.where(largo["fecha"] < pd.Timestamp("2014-01-01"),
                                  "mensual", "quincenal")

largo = (largo.drop(columns=["fecha_str"])
              .sort_values(["CVE_CONCATENADA", "fecha"])
              .reset_index(drop=True))

display(largo.head())
print(f"\nFilas en formato largo: {len(largo):,}")
```

```python
#| label: armonizacion-mensual

tmp = largo.copy()
tmp["anio_mes"] = tmp["fecha"].values.astype("datetime64[M]")

mensual = (tmp.groupby(ID_COLS + ["anio_mes"], as_index=False)
              .agg(nivel_severidad_max=("nivel_severidad", "max"),
                   n_lecturas=("nivel_severidad", "size"),
                   n_lecturas_validas=("nivel_severidad", "count"))
              .rename(columns={"anio_mes": "fecha"}))

inv_map = {0: "Sin sequía", 1: "D0", 2: "D1", 3: "D2", 4: "D3", 5: "D4"}
mensual["categoria_max"] = mensual["nivel_severidad_max"].map(inv_map)
mensual = mensual.sort_values(["CVE_CONCATENADA", "fecha"]).reset_index(drop=True)

display(mensual.head())
print(f"\nFilas en formato mensual armonizado: {len(mensual):,}")
```

```python
#| label: fig-serie-anual
#| fig-cap: Porcentaje de municipios en sequía (D0+) por año, 2003-2025

mensual["anio"] = mensual["fecha"].dt.year
pct_anual = (mensual[mensual["anio"] < 2026]
             .groupby("anio")
             .apply(lambda g: (g["nivel_severidad_max"] > 0).mean() * 100))

fig, ax = plt.subplots()
ax.plot(pct_anual.index, pct_anual.values, marker="o")
ax.set_xlabel("Año")
ax.set_ylabel("% de municipios en sequía")
ax.set_ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
#| label: tbl-distribucion
#| tbl-cap: Distribución de categorías en la serie de resolución nativa

display(largo.loc[~largo["es_hueco_publicacion"], "categoria"]
             .value_counts()
             .rename("frecuencia")
             .to_frame())
```

```python
#| label: guardado

largo.to_parquet("../data/processed/sequia_long.parquet", index=False)
mensual.drop(columns="anio").to_parquet("../data/processed/sequia_mensual.parquet", index=False)

print("Archivos guardados en data/processed/")
```
