```python
import pandas as pd
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

# 1. Cargar datos
df = pd.read_csv("../MunicipiosSequiaCSV.csv")

# 2. Separar columnas geográficas de las columnas temporales
columnas_base = ['CVE_CONCATENADA', 'CVE_ENT', 'CVE_MUN', 'NOMBRE_MUN', 'ENTIDAD', 'ORG_CUENCA*', 'CLV_OC', 'CON_CUENCA', 'CVE_CONC']
columnas_fecha = [col for col in df.columns if col not in columnas_base]

# 3. Transformar de formato ancho a largo (Melt)
df_long = df.melt(id_vars=columnas_base, value_vars=columnas_fecha, 
                  var_name='Fecha_Texto', value_name='Categoria_Sequia')

# 4. Manejo de nulos: Los vacíos indican ausencia de sequía
df_long['Categoria_Sequia'] = df_long['Categoria_Sequia'].fillna('Sin Sequía')

# 5. Estandarización de fechas ultrarrápida (Mapeo de únicos)
meses = {'ene': '01', 'feb': '02', 'mar': '03', 'abr': '04', 'may': '05', 'jun': '06', 
         'jul': '07', 'ago': '08', 'sep': '09', 'oct': '10', 'nov': '11', 'dic': '12'}

def parse_date(date_str):
    try:
        dia, mes, anio = date_str.split('-')
        return pd.to_datetime(f'20{anio}-{meses[mes.lower()]}-{dia}')
    except:
        return pd.NaT

# EL TRUCO: Solo calculamos las fechas únicas (ej. 500 cálculos en lugar de 1.5 millones)
fechas_unicas = df_long['Fecha_Texto'].unique()
mapa_fechas = {fecha: parse_date(fecha) for fecha in fechas_unicas}

# Asignamos el diccionario mapeado a la columna completa instantáneamente
df_long['Fecha'] = df_long['Fecha_Texto'].map(mapa_fechas)

# 6. Crear un índice numérico de severidad para futuros modelos (0 a 5)
mapa_sequia = {'Sin Sequía': 0, 'D0': 1, 'D1': 2, 'D2': 3, 'D3': 4, 'D4': 5}
df_long['Nivel_Num'] = df_long['Categoria_Sequia'].map(mapa_sequia)

# Limpiar columnas innecesarias y ordenar
df_clean = df_long.drop(columns=['Fecha_Texto']).sort_values(['CVE_CONCATENADA', 'Fecha'])

print(f"Estructura final: {df_clean.shape[0]:,} registros.")
df_clean.head()
```

```python
# Contar municipios por categoría en cada fecha
evolucion = df_clean.groupby(['Fecha', 'Categoria_Sequia']).size().reset_index(name='Conteo')

# Colores oficiales del Monitor de Sequía
orden = ['Sin Sequía', 'D0', 'D1', 'D2', 'D3', 'D4']
colores = {'Sin Sequía': '#e0e0e0', 'D0': '#ffff00', 'D1': '#fcd37f', 
           'D2': '#ffaa00', 'D3': '#e60000', 'D4': '#730000'}

# Gráfica de área apilada interactiva
# Gráfica de área apilada interactiva
fig = px.area(evolucion, x='Fecha', y='Conteo', 
              color='Categoria_Sequia',
              category_orders={'Categoria_Sequia': orden},
              color_discrete_map=colores)

fig.update_layout(
    title_text='Proporción de Municipios por Nivel de Severidad de Sequía',
    xaxis_title='Año',
    yaxis_title='Número de Municipios',
    template='plotly_white',
    hovermode='x unified',
    legend_title_text='Nivel de Sequía',
    margin=dict(l=20, r=20, t=50, b=20)
)

# Mostrar la gráfica en Quarto
fig.show()
```
